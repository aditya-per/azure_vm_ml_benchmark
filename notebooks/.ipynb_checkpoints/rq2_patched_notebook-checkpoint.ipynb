```python
def tune_threshold_by_f1(y_true, scores, higher_score_more_anomalous=True):
    import numpy as np
    y_true=np.asarray(y_true,dtype=int)
    scores=np.asarray(scores,dtype=float)
    valid=np.isfinite(scores)
    y=y_true[valid]
    scores=scores[valid]
    working=scores if higher_score_more_anomalous else -scores
    precision, recall, thresholds = precision_recall_curve(y, working)
    f1=(2*precision[:-1]*recall[:-1])/(precision[:-1]+recall[:-1]+1e-12)
    idx=int(np.nanargmax(f1))
    thr=float(thresholds[idx])
    return {'threshold':thr if higher_score_more_anomalous else -thr,'f1':float(f1[idx]),'precision':float(precision[idx]),'recall':float(recall[idx]),'average_precision':float(average_precision_score(y,working))}
```

```python
# PATCH 7: timestamp-aligned LSTM labels
def build_lstm_sequences(df,value_cols,seq_len=24):
    X=[]; y=[]; keys=[]
    for vm,group in df.sort_values(['vm_id','hour_index']).groupby('vm_id'):
        values=group[value_cols].to_numpy(dtype='float32')
        labels=group['is_anomaly'].to_numpy(dtype='int32')
        hours=group['hour_index'].to_numpy(dtype='int32')
        if len(group)<seq_len: continue
        for end in range(seq_len-1,len(group)):
            start=end-seq_len+1
            X.append(values[start:end+1])
            y.append(int(labels[end]))
            keys.append((int(vm),int(hours[end])))
    return np.asarray(X,dtype='float32'), np.asarray(y,dtype='int32'), keys
```

```python
def evaluate_event_level(scored_df,event_df,pred_col):
    import pandas as pd
    work=scored_df.sort_values(['vm_id','hour_index']).copy()
    prev=work.groupby('vm_id')[pred_col].shift(1,fill_value=0)
    work['_start']=(work[pred_col].eq(1)&prev.eq(0))
    work['_eid']=work.groupby('vm_id')['_start'].cumsum()
    predicted=work[work[pred_col]==1]
    if len(predicted):
        pred_events=predicted.groupby(['vm_id','_eid']).agg(overlaps_true=('is_anomaly','max')).reset_index()
        pe=len(pred_events); correct=int(pred_events['overlaps_true'].sum())
    else:
        pe=0; correct=0
    detected=work.loc[(work['is_anomaly']==1)&(work[pred_col]==1),'anomaly_event_id'].nunique()
    true_count=len(event_df)
    p=correct/pe if pe else 0
    r=detected/true_count if true_count else 0
    f1=2*p*r/(p+r) if p+r else 0
    return {'event_precision':p,'event_recall':r,'event_f1':f1}
```
