# 03_RQ2_Anomaly_Detection

## Research Question 2: Anomaly Detection

**RQ2:** Can unsupervised methods such as Isolation Forest and LSTM Autoencoder detect anomalous VM behaviour, including spikes, dips, and dead VMs, with higher F1 score than a static three-sigma thresholding rule, using synthetic anomaly injection as ground truth?

**Hypothesis H2:** LSTM Autoencoder achieves F1 greater than or equal to 0.80 and outperforms three-sigma thresholding by at least 20 percentage points in F1 on injected anomalies of moderate magnitude.

This notebook builds the RQ2 anomaly detection workflow using hourly Azure VM CPU utilization data.

Models included:

- Rolling three-sigma threshold baseline
- Isolation Forest
- optional Local Outlier Factor if needed
- optional LSTM Autoencoder if TensorFlow/Keras is available

**Scope note:** The anomaly detection design uses CPU utilization only. Memory, disk, network, and application telemetry are not available in the public trace, so detected anomalies should be interpreted as CPU-utilization anomalies rather than complete service-health incidents.

`data/processed/hourly_n2000.parquet`

Update `DATA_PATH` if your dataset is stored elsewhere.

```python
import os
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler

pd.set_option('display.max_columns', 80)
pd.set_option('display.width', 140)
plt.rcParams['figure.figsize'] = (12, 5)

RANDOM_STATE = 42
DATA_PATH = '../data/processed/hourly_n2000.parquet'
OUTPUT_DIR = '../data/reports/rq2_anomaly_detection'

if not os.path.exists(DATA_PATH):
    alt_path = 'data/processed/hourly_n2000.parquet'
    if os.path.exists(alt_path):
        DATA_PATH = alt_path

if not os.path.exists(OUTPUT_DIR):
    OUTPUT_DIR = 'data/reports/rq2_anomaly_detection'

os.makedirs(OUTPUT_DIR, exist_ok=True)

print('DATA_PATH:', DATA_PATH)
print('OUTPUT_DIR:', OUTPUT_DIR)
```

The primary anomaly detection target is `avg_cpu_mean`, with `avg_cpu_max`, `avg_cpu_min`, and `avg_cpu_std` used as supporting CPU-utilization signals.

```python
df = pd.read_parquet(DATA_PATH)
df = df.sort_values(['vm_id', 'hour_index']).reset_index(drop=True)

print('Shape:', df.shape)
print('Unique VMs:', df['vm_id'].nunique())
print('Unique hours:', df['hour_index'].nunique())
df.head()
```

```python
required_cols = ['vm_id', 'hour_index', 'avg_cpu_mean', 'avg_cpu_max', 'avg_cpu_min', 'avg_cpu_std', 'hour_of_day', 'day_index', 'day_of_cycle']
missing_cols = [c for c in required_cols if c not in df.columns]
if missing_cols:
    raise ValueError(f'Missing required columns: {missing_cols}')

print(df[required_cols].isna().sum())
print('Duplicate VM-hour rows:', df.duplicated(['vm_id', 'hour_index']).sum())
```

## 3. Define Train, Validation, and Test Windows

Anomaly detection models are trained on the earlier part of the time series and evaluated on synthetic anomalies injected into the test period.

- Training: hours 0 through 503
- Validation: hours 504 through 599
- Test: hours 600 through 719

Synthetic anomaly injection is used because the public dataset does not provide verified anomaly labels.

```python
TRAIN_END = 503
VAL_END = 599
TEST_END = 719

train_df_raw = df[df['hour_index'] <= TRAIN_END].copy()
val_df_raw = df[(df['hour_index'] > TRAIN_END) & (df['hour_index'] <= VAL_END)].copy()
test_df_raw = df[(df['hour_index'] > VAL_END) & (df['hour_index'] <= TEST_END)].copy()

print('Train:', train_df_raw.shape)
print('Validation:', val_df_raw.shape)
print('Test:', test_df_raw.shape)
```

## 4. Synthetic Anomaly Injection Design

Synthetic anomalies are injected into the test set to create known ground truth labels. Three anomaly types are included:

1. **Spike anomalies:** sudden high CPU jumps.
2. **Dip anomalies:** sudden sharp decreases.
3. **Dead VM anomalies:** near-zero CPU for a sustained segment.

This design allows Precision, Recall, F1, and Mean Time to Detect to be measured. Synthetic injection is a limitation because real incident labels are unavailable.

```python
def inject_synthetic_anomalies(input_df, anomaly_rate=0.03, min_duration=2, max_duration=8, random_state=42):
    rng = np.random.default_rng(random_state)
    data = input_df.sort_values(['vm_id', 'hour_index']).copy()
    data['is_anomaly'] = 0
    data['anomaly_type'] = 'normal'
    data['avg_cpu_mean_original'] = data['avg_cpu_mean']

    vm_ids = data['vm_id'].unique()
    n_vms_to_inject = max(1, int(len(vm_ids) * anomaly_rate))
    selected_vms = rng.choice(vm_ids, size=n_vms_to_inject, replace=False)
    anomaly_types = ['spike', 'dip', 'dead_vm']

    for vm in selected_vms:
        vm_idx = data.index[data['vm_id'] == vm].to_numpy()
        if len(vm_idx) < max_duration + 2:
            continue

        duration = int(rng.integers(min_duration, max_duration + 1))
        start_pos = int(rng.integers(0, len(vm_idx) - duration))
        selected_idx = vm_idx[start_pos:start_pos + duration]
        anomaly_type = rng.choice(anomaly_types)

        current_values = data.loc[selected_idx, 'avg_cpu_mean'].values

        if anomaly_type == 'spike':
            injected_values = np.maximum(current_values, np.minimum(100, current_values + rng.uniform(40, 80)))
        elif anomaly_type == 'dip':
            injected_values = np.maximum(0, current_values * rng.uniform(0.05, 0.25))
        else:
            injected_values = rng.uniform(0, 1, size=len(selected_idx))

        data.loc[selected_idx, 'avg_cpu_mean'] = injected_values
        data.loc[selected_idx, 'avg_cpu_max'] = np.maximum(data.loc[selected_idx, 'avg_cpu_max'], injected_values)
        data.loc[selected_idx, 'avg_cpu_min'] = np.minimum(data.loc[selected_idx, 'avg_cpu_min'], injected_values)
        data.loc[selected_idx, 'is_anomaly'] = 1
        data.loc[selected_idx, 'anomaly_type'] = anomaly_type

    return data

test_df = inject_synthetic_anomalies(test_df_raw, anomaly_rate=0.08, random_state=RANDOM_STATE)

print('Injected anomaly rows:', test_df['is_anomaly'].sum())
print('Injected anomaly rate:', test_df['is_anomaly'].mean())
print(test_df['anomaly_type'].value_counts())
```

```python
# Visualize a few injected anomalies
anomaly_vms = test_df.loc[test_df['is_anomaly'] == 1, 'vm_id'].drop_duplicates().head(4).tolist()

for vm in anomaly_vms:
    temp = test_df[test_df['vm_id'] == vm].sort_values('hour_index')
    plt.figure(figsize=(14, 4))
    plt.plot(temp['hour_index'], temp['avg_cpu_mean_original'], label='Original CPU', alpha=0.8)
    plt.plot(temp['hour_index'], temp['avg_cpu_mean'], label='After injection', alpha=0.8)
    anomaly_points = temp[temp['is_anomaly'] == 1]
    plt.scatter(anomaly_points['hour_index'], anomaly_points['avg_cpu_mean'], color='red', label='Injected anomalies')
    plt.title(f'Synthetic Anomaly Injection Example: vm_id={vm}')
    plt.xlabel('Hour index')
    plt.ylabel('CPU utilization (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

## 5. Feature Engineering for Anomaly Detection

Features are designed to capture current utilization, recent trend, volatility, and seasonality.

Important implementation detail: lag and rolling features are engineered on the full ordered VM history before splitting back into train, validation, and test sets. This preserves past context for validation and test periods, especially for lag-168 and rolling-168 features. Since lag and rolling calculations use only prior observations through `shift(1)`, this avoids future data leakage while preventing validation/test rows from being dropped due to missing long-history features.


```python
# Combine train, validation, and injected test data before feature engineering.
# This preserves historical context needed for lag_168 and rolling_168 features.
# Because all lag/rolling features use shift(1), only past values are used, so this does not introduce future leakage.

train_df_raw_labeled = train_df_raw.copy()
val_df_raw_labeled = val_df_raw.copy()
test_df_labeled = test_df.copy()

train_df_raw_labeled['is_anomaly'] = 0
train_df_raw_labeled['anomaly_type'] = 'normal'

val_df_raw_labeled['is_anomaly'] = 0
val_df_raw_labeled['anomaly_type'] = 'normal'

# test_df already contains injected anomaly labels and anomaly_type
full_labeled_df = pd.concat(
    [train_df_raw_labeled, val_df_raw_labeled, test_df_labeled],
    axis=0,
    ignore_index=True
)

full_labeled_df = full_labeled_df.sort_values(['vm_id', 'hour_index']).reset_index(drop=True)

print('Full labeled data shape:', full_labeled_df.shape)
print('Anomaly rows in full data:', full_labeled_df['is_anomaly'].sum())
print(full_labeled_df[['vm_id', 'hour_index', 'is_anomaly', 'anomaly_type']].head())

# Engineer features on the full ordered VM history, then split afterward.
features_all_df = add_anomaly_features(full_labeled_df)

feature_cols = [
    c for c in features_all_df.columns
    if c.startswith('lag_') or c.startswith('roll_') or c in [
        'avg_cpu_mean', 'avg_cpu_max', 'avg_cpu_min', 'avg_cpu_std', 'readings_count',
        'hour_sin', 'hour_cos', 'day_cycle_sin', 'day_cycle_cos',
        'hourly_range', 'cpu_change_1h', 'abs_cpu_change_1h', 'z_roll_24', 'z_roll_168'
    ]
]
feature_cols = [c for c in feature_cols if c in features_all_df.columns]

train_model_df = features_all_df[
    features_all_df['hour_index'] <= TRAIN_END
].dropna(subset=feature_cols).copy()

val_model_df = features_all_df[
    (features_all_df['hour_index'] > TRAIN_END) &
    (features_all_df['hour_index'] <= VAL_END)
].dropna(subset=feature_cols).copy()

test_model_df = features_all_df[
    (features_all_df['hour_index'] > VAL_END) &
    (features_all_df['hour_index'] <= TEST_END)
].dropna(subset=feature_cols + ['is_anomaly']).copy()

print('Feature count:', len(feature_cols))
print('Train rows:', train_model_df.shape)
print('Validation rows:', val_model_df.shape)
print('Test rows:', test_model_df.shape)
print('Test anomaly rows:', int(test_model_df['is_anomaly'].sum()))
print('Test anomaly rate:', test_model_df['is_anomaly'].mean())

if len(val_model_df) == 0 or len(test_model_df) == 0:
    raise ValueError('Validation or test rows are still zero. Check split windows or feature construction.')
```

## 6. Evaluation Metrics

The main classification metrics are:

- Precision
- Recall
- F1 score
- False positive rate
- Mean Time to Detect, or MTTD, for multi-hour injected anomaly segments

False positives are important because excessive operational alerts reduce practitioner trust.

```python
def anomaly_metrics(y_true, y_pred, model_name):
    y_true = np.asarray(y_true).astype(int)
    y_pred = np.asarray(y_pred).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    fpr = fp / (fp + tn + 1e-9)

    return {
        'model': model_name,
        'TP': tp,
        'FP': fp,
        'TN': tn,
        'FN': fn,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'False Positive Rate': fpr
    }

results = []
```

## 7. Baseline: Rolling Three-Sigma Threshold

The baseline flags an anomaly when the hourly CPU value is more than three rolling standard deviations away from the rolling 24-hour mean.

```python
def three_sigma_detector(data, z_col='z_roll_24', threshold=3.0):
    pred = (data[z_col].abs() >= threshold).astype(int)
    return pred

test_model_df['pred_three_sigma'] = three_sigma_detector(test_model_df, threshold=3.0)
results.append(anomaly_metrics(test_model_df['is_anomaly'], test_model_df['pred_three_sigma'], 'Rolling three-sigma'))

pd.DataFrame(results)
```

## 8. Isolation Forest

Isolation Forest is trained on the training period, which is treated as mostly normal. The contamination level is set conservatively because synthetic anomalies are injected only into the test period. Validation rows are retained for diagnostics and future threshold tuning.


```python
scaler = StandardScaler()
X_train = scaler.fit_transform(train_model_df[feature_cols])
X_test = scaler.transform(test_model_df[feature_cols])

iso_model = IsolationForest(
    n_estimators=300,
    contamination=0.03,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

iso_model.fit(X_train)

# sklearn returns -1 for anomaly and 1 for normal
iso_raw_pred = iso_model.predict(X_test)
test_model_df['pred_isolation_forest'] = (iso_raw_pred == -1).astype(int)

results.append(anomaly_metrics(test_model_df['is_anomaly'], test_model_df['pred_isolation_forest'], 'Isolation Forest'))

pd.DataFrame(results).sort_values('F1', ascending=False)
```

## 9. Optional Local Outlier Factor

This section is optional. LOF is useful as another unsupervised baseline but can be slower on large datasets. To keep runtime reasonable, it is evaluated on the test feature matrix using novelty detection.

```python
RUN_LOF = False

if RUN_LOF:
    from sklearn.neighbors import LocalOutlierFactor

    lof_model = LocalOutlierFactor(n_neighbors=35, contamination=0.03, novelty=True, n_jobs=-1)
    lof_model.fit(X_train)
    lof_raw_pred = lof_model.predict(X_test)
    test_model_df['pred_lof'] = (lof_raw_pred == -1).astype(int)
    results.append(anomaly_metrics(test_model_df['is_anomaly'], test_model_df['pred_lof'], 'Local Outlier Factor'))

pd.DataFrame(results).sort_values('F1', ascending=False)
```

## 10. Optional LSTM Autoencoder

The LSTM Autoencoder learns normal CPU patterns and flags sequences with high reconstruction error. This section runs only if TensorFlow/Keras is installed.

To reduce runtime, the model is trained on a VM sample by default.

```python
LSTM_VM_SAMPLE_SIZE = 200
LOOKBACK = 48
EPOCHS = 5
BATCH_SIZE = 128

try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import LSTM, Dense, RepeatVector, TimeDistributed, Dropout
    from tensorflow.keras.callbacks import EarlyStopping
    TF_AVAILABLE = True
    print('TensorFlow available:', tf.__version__)
except Exception:
    TF_AVAILABLE = False
    print('TensorFlow/Keras not available. LSTM Autoencoder section will be skipped.')
```

```python
def make_autoencoder_sequences(input_df, vm_ids, lookback=48, target_col='avg_cpu_mean'):
    X = []
    meta = []
    for vm in vm_ids:
        temp = input_df[input_df['vm_id'] == vm].sort_values('hour_index')
        series = (temp[target_col].values.astype('float32') / 100.0)
        hours = temp['hour_index'].values
        labels = temp['is_anomaly'].values if 'is_anomaly' in temp.columns else np.zeros(len(temp), dtype=int)

        for i in range(lookback, len(series)):
            X.append(series[i - lookback:i])
            meta.append({
                'vm_id': vm,
                'hour_index': hours[i],
                'is_anomaly': int(labels[i])
            })

    X = np.array(X).reshape(-1, lookback, 1)
    meta = pd.DataFrame(meta)
    return X, meta

if TF_AVAILABLE:
    rng = np.random.default_rng(RANDOM_STATE)
    selected_vms = rng.choice(df['vm_id'].unique(), size=min(LSTM_VM_SAMPLE_SIZE, df['vm_id'].nunique()), replace=False)

    train_lstm_source = train_df_raw[train_df_raw['vm_id'].isin(selected_vms)].copy()
    test_lstm_source = test_df[test_df['vm_id'].isin(selected_vms)].copy()

    X_ae_train, train_meta = make_autoencoder_sequences(train_lstm_source.assign(is_anomaly=0), selected_vms, LOOKBACK)
    X_ae_test, test_meta = make_autoencoder_sequences(test_lstm_source, selected_vms, LOOKBACK)

    print('Autoencoder train:', X_ae_train.shape)
    print('Autoencoder test:', X_ae_test.shape)
```

```python
if TF_AVAILABLE:
    tf.keras.utils.set_random_seed(RANDOM_STATE)

    ae_model = Sequential([
        LSTM(32, activation='tanh', input_shape=(LOOKBACK, 1), return_sequences=False),
        Dropout(0.2),
        RepeatVector(LOOKBACK),
        LSTM(32, activation='tanh', return_sequences=True),
        TimeDistributed(Dense(1))
    ])

    ae_model.compile(optimizer='adam', loss='mse')
    early_stop = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)

    history = ae_model.fit(
        X_ae_train, X_ae_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[early_stop],
        verbose=1
    )

    train_recon = ae_model.predict(X_ae_train, verbose=0)
    train_errors = np.mean(np.square(X_ae_train - train_recon), axis=(1, 2))
    threshold = np.percentile(train_errors, 97)

    test_recon = ae_model.predict(X_ae_test, verbose=0)
    test_errors = np.mean(np.square(X_ae_test - test_recon), axis=(1, 2))

    test_meta['reconstruction_error'] = test_errors
    test_meta['pred_lstm_autoencoder'] = (test_errors >= threshold).astype(int)

    results.append(anomaly_metrics(test_meta['is_anomaly'], test_meta['pred_lstm_autoencoder'], 'LSTM Autoencoder sampled VMs'))

pd.DataFrame(results).sort_values('F1', ascending=False)
```

## 11. Mean Time to Detect, MTTD

MTTD measures detection latency for injected anomaly segments. It is calculated as the average number of hours between anomaly onset and first detection within the same VM anomaly segment.

```python
def calculate_mttd(data, pred_col):
    temp = data.sort_values(['vm_id', 'hour_index']).copy()
    anomaly_segments = []

    for vm, vm_data in temp.groupby('vm_id'):
        vm_data = vm_data.sort_values('hour_index').copy()
        is_anom = vm_data['is_anomaly'].values
        hours = vm_data['hour_index'].values
        preds = vm_data[pred_col].values

        in_seg = False
        start_idx = None
        for i, flag in enumerate(is_anom):
            if flag == 1 and not in_seg:
                in_seg = True
                start_idx = i
            end_condition = (flag == 0 and in_seg) or (i == len(is_anom) - 1 and in_seg)
            if end_condition:
                end_idx = i - 1 if flag == 0 else i
                segment_pred = preds[start_idx:end_idx + 1]
                segment_hours = hours[start_idx:end_idx + 1]
                if segment_pred.sum() > 0:
                    first_detection_hour = segment_hours[np.argmax(segment_pred == 1)]
                    latency = first_detection_hour - segment_hours[0]
                    anomaly_segments.append(latency)
                else:
                    anomaly_segments.append(np.nan)
                in_seg = False

    if len(anomaly_segments) == 0:
        return np.nan, 0, 0

    detected = np.sum(~pd.isna(anomaly_segments))
    total = len(anomaly_segments)
    mttd = np.nanmean(anomaly_segments)
    return mttd, detected, total

mttd_rows = []
for pred_col, model_name in [
    ('pred_three_sigma', 'Rolling three-sigma'),
    ('pred_isolation_forest', 'Isolation Forest')
]:
    mttd, detected, total = calculate_mttd(test_model_df, pred_col)
    mttd_rows.append({'model': model_name, 'MTTD_hours': mttd, 'segments_detected': detected, 'total_segments': total})

mttd_df = pd.DataFrame(mttd_rows)
mttd_df
```
